```python
import pandas as pd
import numpy as np

from sklearn.datasets import load_iris
from sklearn.preprocessing import LabelEncoder

print("Libraries imported successfully.")
```

```python
iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["species"] = iris.target_names[iris.target]

print("Iris dataset loaded successfully.")
```

```python
print("First 5 rows of the dataset:")
df.head()
```

```python
print("Last 5 rows of the dataset:")
df.tail()
```

```python
print("Dataset Shape:", df.shape)

print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
```

```python
print("Dataset Information:")
df.info()

print("\nData Types:")
print(df.dtypes)
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())
```

```python
duplicate_count = df.duplicated().sum()

print("Number of Duplicate Rows:", duplicate_count)
```

```python
duplicates = df[df.duplicated()]

if len(duplicates) == 0:
    print("No duplicate records found.")
else:
    print("Duplicate records found:")
    display(duplicates)
```

```python
df_clean = df.drop_duplicates().copy()

print("Original Dataset Shape:", df.shape)
print("Dataset Shape After Removing Duplicates:", df_clean.shape)
```

```python
print("Missing Values After Cleaning:")
print(df_clean.isnull().sum())

print("\nTotal Missing Values:",
      df_clean.isnull().sum().sum())
```

```python
print("Statistical Summary:")
df_clean.describe()
```

```python
print("Number of Flowers in Each Species:")
print(df_clean["species"].value_counts())
```

```python
X = df_clean.drop("species", axis=1)
y = df_clean["species"]

print("Feature Columns:")
print(X.columns.tolist())

print("\nTarget Variable:")
print("species")

print("\nX Shape:", X.shape)
print("y Shape:", y.shape)
```

```python
label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print("Species Classes:")
print(label_encoder.classes_)

print("\nFirst 10 Encoded Target Values:")
print(y_encoded[:10])
```

```python
df_clean["species_encoded"] = y_encoded

print("Dataset with Encoded Target:")
df_clean.head()
```

```python
print("=" * 50)
print("FINAL PREPROCESSED DATASET")
print("=" * 50)

print("Rows:", df_clean.shape[0])
print("Columns:", df_clean.shape[1])
print("Missing Values:", df_clean.isnull().sum().sum())
print("Duplicate Rows:", df_clean.duplicated().sum())

print("\nFeature Shape:", X.shape)
print("Target Shape:", y_encoded.shape)

print("\nPreprocessing completed successfully.")
```

Data Collection and Preprocessing
The Iris dataset was loaded using the Scikit-learn library and converted into a Pandas DataFrame for analysis. The dataset contains four numerical input features: sepal length, sepal width, petal length and petal width. The target variable represents the Iris flower species.
The first and last five records were inspected using head() and tail(). The shape, data types and general information of the dataset were also examined. Missing values and duplicate records were checked to identify potential data-quality issues. Duplicate records, if present, were removed to create a clean dataset.
The input features were separated from the target variable. Since the target originally contains categorical species names, Label Encoding was used to convert Setosa, Versicolor and Virginica into numerical class labels suitable for machine-learning algorithms.
The resulting dataset is now cleaned, structured and ready for exploratory data analysis and model development.
