```python
import pandas as pd
import numpy as np

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

print("Libraries imported successfully.")
```

```python
iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["species"] = iris.target_names[iris.target]

print("Iris dataset loaded successfully.")
print("Original shape:", df.shape)

df.head()
```

```python
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates().copy()

print("Shape after cleaning:", df.shape)
```

```python
X = df.drop("species", axis=1)
y = df["species"]

print("Feature Variables:")
print(X.columns.tolist())

print("\nTarget Variable:")
print("species")

print("\nX Shape:", X.shape)
print("y Shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Features Shape:", X_train.shape)
print("Testing Features Shape:", X_test.shape)

print("Training Samples:", len(X_train))
print("Testing Samples:", len(X_test))
```

```python
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

print("Random Forest Classifier created successfully.")
print("Number of Trees:", model.n_estimators)
```

Model Selection: Random Forest Classifier
Random Forest was selected because the Iris flower problem is a multiclass classification problem. Random Forest combines multiple decision trees and uses their results to produce a final prediction.
It is suitable for this dataset because it can learn both simple and non-linear relationships between numerical features such as sepal length, sepal width, petal length and petal width. It also works directly with multiclass target variables and does not require feature scaling.
Random Forest is generally robust, easy to train and performs effectively on small structured datasets such as Iris.

```python
model.fit(
    X_train,
    y_train
)

print("Model training completed successfully.")
```

```python
print("Model Type:")
print(type(model).__name__)

print("\nNumber of Trees:")
print(model.n_estimators)

print("\nClasses learned by the model:")
print(model.classes_)
```

```python
predictions = model.predict(X_test)

print("Predictions generated successfully.")
print("Number of Predictions:", len(predictions))
```

```python
prediction_results = pd.DataFrame({
    "Actual Species": y_test.values,
    "Predicted Species": predictions
})

print("Sample Predictions:")

prediction_results.head(10)
```

```python
new_flower = pd.DataFrame(
    [[5.1, 3.5, 1.4, 0.2]],
    columns=X.columns
)

new_prediction = model.predict(new_flower)

print("New Flower Measurements:")
print(new_flower)

print("\nPredicted Species:")
print(new_prediction[0])
```

```python
feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print("Feature Importance:")
feature_importance
```

Model Development and Training Summary
The cleaned Iris dataset was divided into input features and a target variable. Sepal length, sepal width, petal length and petal width were used as the input features, while the Iris species was selected as the target.
The dataset was divided into 80% training data and 20% testing data. Stratified sampling was used to maintain a similar distribution of the three flower species in both datasets.
A Random Forest Classifier containing 100 decision trees was selected for the classification task. The model was trained using the training dataset, allowing it to learn relationships between flower measurements and their corresponding species.
After training, predictions were generated for the unseen test data. The trained model was also used to classify a new flower sample. Random Forest feature importance showed which flower measurements contributed most strongly to classification.
