```python
import matplotlib.pyplot as plt

print("Matplotlib is working!")
```

```python
import pandas as pd

print("Pandas is working!")
```

```python
df = pd.read_csv("ai_model_performance.csv")

df
```

```python
#Bar graph
plt.figure(figsize=(8, 5))

plt.bar(df["Model"], df["Accuracy"])

plt.title("Accuracy of AI Models")
plt.xlabel("AI Model")
plt.ylabel("Accuracy")
plt.xticks(rotation=45)

plt.show()
```

 Bar graph :The bar chart compares the accuracy of different AI models. The Neural Network has the highest accuracy, while Linear Regression has the lowest accuracy.

```python
#Line Chart
plt.figure(figsize=(8, 5))

plt.plot(df["Model"], df["Training_Time"], marker="o")

plt.title("Training Time of AI Models")
plt.xlabel("AI Model")
plt.ylabel("Training Time")

plt.xticks(rotation=45)

plt.show()
```

Line Chart:The line chart compares the training time of different AI models. The Neural Network takes the longest training time, while Linear Regression takes the shortest.

```python
#Histogram
plt.figure(figsize=(8, 5))

plt.hist(df["Accuracy"], bins=5)

plt.title("Distribution of AI Model Accuracy")
plt.xlabel("Accuracy")
plt.ylabel("Frequency")

plt.show()
```

Histogram:The histogram shows how the accuracy values of the AI models are distributed. Most of the models have accuracy values between 78 and 94.

```python
#Scatter Plot
plt.figure(figsize=(8, 5))

plt.scatter(df["Training_Time"], df["Accuracy"])

plt.title("Training Time vs Accuracy")
plt.xlabel("Training Time")
plt.ylabel("Accuracy")

plt.show()
```

Scatter Plot:The scatter plot shows the relationship between training time and accuracy. In this dataset, models with higher training time generally have higher accuracy.
