```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("student_performance.csv")

df
```

```python
print("Dataset Information:")
df.info()
```

```python
print("Column Names:")
print(df.columns)

print("\nImportant Columns:")
print("Student_Name - identifies each student")
print("Marks - used to analyze student performance")
print("Attendance - used to compare attendance with marks")
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
average_before_cleaning = df["Marks"].mean()

print("Average Marks before filling missing value:",
      round(average_before_cleaning, 2))

df["Marks"] = df["Marks"].fillna(average_before_cleaning)

print("\nMissing values after filling:")
print(df.isnull().sum())
```

```python
df = df.drop_duplicates()

print("Number of duplicate records after cleaning:")
print(df.duplicated().sum())
```

```python
print("Cleaned Dataset:")
df
```

```python
average_marks = df["Marks"].mean()

print("Average Marks:", round(average_marks, 2))
```

```python
highest_marks = df["Marks"].max()

print("Highest Marks:", highest_marks)
```

```python
lowest_marks = df["Marks"].min()

print("Lowest Marks:", lowest_marks)
```

```python
pass_mark = 40

passed_students = df[df["Marks"] >= pass_mark]

print("Number of Students Passed:", len(passed_students))
```

```python
failed_students = df[df["Marks"] < pass_mark]

print("Number of Students Failed:", len(failed_students))
```

```python
print("Passed Students:")
print(passed_students[["Student_Name", "Marks"]])

print("\nFailed Students:")
print(failed_students[["Student_Name", "Marks"]])
```

```python
plt.figure(figsize=(10, 5))

plt.bar(df["Student_Name"], df["Marks"])

plt.title("Student Marks")
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.xticks(rotation=45)

plt.show()
```

The bar chart compares the marks obtained by each student.
It helps identify students with higher and lower academic performance.

```python
pass_count = len(df[df["Marks"] >= 40])
fail_count = len(df[df["Marks"] < 40])

plt.figure(figsize=(6, 6))

plt.pie(
    [pass_count, fail_count],
    labels=["Passed", "Failed"],
    autopct="%1.1f%%"
)

plt.title("Student Pass and Fail Distribution")

plt.show()
```

The pie chart shows the percentage of students who passed and failed.
It provides a quick overview of the overall class performance.

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Attendance"], df["Marks"])

plt.title("Attendance vs Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Marks")

plt.show()
```

The scatter plot shows the relationship between attendance and marks.
It helps us observe whether students with higher attendance tend to achieve better marks.
