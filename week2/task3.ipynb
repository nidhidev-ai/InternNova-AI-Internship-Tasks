```python
import pandas as pd
data = {
    "Model": ["Linear Regression", "Decision Tree", "Random Forest",
              "KNN", "Neural Network", "Random Forest"],

    "Accuracy": [78, 82, None, 80, 94, 89],

    "Training_Time": [2, 5, 10, None, 15, 10],

    "Dataset_Size": [1000, 1500, 2000, 1200, 2500, 2000]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.isnull())
print(df.isnull().sum())
```

```python
df["Accuracy"] = df["Accuracy"].fillna(df["Accuracy"].mean())
df["Training_Time"] = df["Training_Time"].fillna(df["Training_Time"].mean())

print(df)
```

```python
print("Duplicate records:")
print(df[df.duplicated()])
```

```python
df.loc[5] = df.loc[2]

print(df)
```

```python
print("Number of duplicate records:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
print(df)
print(df.dtypes)
```

```python
df["Accuracy"] = df["Accuracy"].astype(float)
df["Training_Time"] = df["Training_Time"].astype(float)

print(df.dtypes)
```

```python
print("Cleaned Dataset:")
print(df)
print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
print("Duplicate records after cleaning:", df.duplicated().sum())
```
