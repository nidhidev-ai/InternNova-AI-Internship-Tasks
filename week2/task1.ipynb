```python
# Creating a 1D NumPy array
import numpy as np
arr1 = np.array([10, 20, 30, 40, 50])

print("1D Array:", arr1)
```

```python
# Creating a 2D NumPy array
arr2 = np.array([[10, 20, 30],
                 [40, 50, 60]])

print("2D Array:")
print(arr2)
```

```python
arr = np.array([10, 20, 30, 40, 50])

print("First element:", arr[0])
print("Third element:", arr[2])
print("Last element:", arr[-1])
```

```python
arr2 = np.array([[10, 20, 30],
                 [40, 50, 60]])

print("Element at row 1, column 2:", arr2[0, 1])
print("Element at row 2, column 3:", arr2[1, 2])
```

```python
arr = np.array([10, 20, 30, 40, 50])

print("First three elements:", arr[0:3])
print("Elements from index 2:", arr[2:])
print("Every second element:", arr[::2])
```

```python
arr2 = np.array([[10, 20, 30],
                 [40, 50, 60]])

print("First row:", arr2[0, :])
print("First two columns:")
print(arr2[0:1, :])
```

```python
arr1 = np.array([10, 20, 30, 40, 50])

arr2 = np.array([[10, 20, 30],
                 [40, 50, 60]])

print("1D Array Shape:", arr1.shape)
print("1D Array Dimensions:", arr1.ndim)

print("2D Array Shape:", arr2.shape)
print("2D Array Dimensions:", arr2.ndim)
```

```python
arr = np.array([10, 20, 30, 40, 50])

print("Addition:", arr + 5)
print("Subtraction:", arr - 5)
print("Multiplication:", arr * 2)
print("Division:", arr / 2)
```

```python
arr = np.array([10, 20, 30, 40, 50])

print("Mean:", np.mean(arr))
print("Maximum:", np.max(arr))
print("Minimum:", np.min(arr))
print("Sum:", np.sum(arr))
```
