```python
import os
import warnings

os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"
warnings.filterwarnings("ignore")

import numpy as np
import tensorflow as tf

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# For reproducible results
np.random.seed(42)
tf.random.set_seed(42)

print("Libraries imported successfully.")
```

```python
# Load Iris dataset
iris = load_iris()

# Features and target
X = iris.data
y = iris.target

print("Dataset loaded successfully.")
print("Dataset shape:", X.shape)
print("Feature names:", iris.feature_names)
print("Target classes:", iris.target_names)
```

```python
print("First 5 feature records:")
print(X[:5])

print("\nFirst 5 target values:")
print(y[:5])
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training feature shape:", X_train.shape)
print("Testing feature shape:", X_test.shape)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
scaler = StandardScaler()

# Learn scaling only from training data
X_train = scaler.fit_transform(X_train)

# Apply the same scaling to test data
X_test = scaler.transform(X_test)

print("Data preprocessing completed.")
print("\nFirst 5 scaled training samples:")
print(X_train[:5])
```

```python
model = tf.keras.Sequential([

    # Input layer: 4 Iris features
    tf.keras.layers.Input(shape=(4,)),

    # First hidden layer
    tf.keras.layers.Dense(
        units=16,
        activation="relu"
    ),

    # Second hidden layer
    tf.keras.layers.Dense(
        units=8,
        activation="relu"
    ),

    # Output layer: 3 flower classes
    tf.keras.layers.Dense(
        units=3,
        activation="softmax"
    )
])

print("Neural network created successfully.")
```

```python
model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("Model compiled successfully.")
```

```python
history = model.fit(
    X_train,
    y_train,
    epochs=50,
    batch_size=8,
    validation_split=0.20,
    verbose=1
)
```
